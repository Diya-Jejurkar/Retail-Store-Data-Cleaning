```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/retail_store_sales_cleaned.csv")

print("Cleaned dataset loaded successfully!")
df
```

```python
print("FIRST 5 ROWS:")
df.head()
```

```python
print("DATASET SHAPE:")
print(df.shape)
```

```python
print("DATASET INFORMATION:")
df.info()
```

```python
print("STATISTICAL SUMMARY:")
df.describe()
```

```python
print("TOTAL SALES:")
print(df["Total Spent"].sum())

print("\nAVERAGE SALES:")
print(df["Total Spent"].mean())

print("\nMINIMUM SALE:")
print(df["Total Spent"].min())

print("\nMAXIMUM SALE:")
print(df["Total Spent"].max())
```

```python
category_sales = df.groupby("Category")["Total Spent"].sum().sort_values(ascending=False)

print("SALES BY CATEGORY:")
print(category_sales)
```

```python
category_sales.plot(kind="bar", figsize=(10,5))

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
payment_counts = df["Payment Method"].value_counts()

print("TRANSACTIONS BY PAYMENT METHOD:")
print(payment_counts)
```

```python
payment_counts.plot(kind="bar", figsize=(8,5))

plt.title("Transactions by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()
```

```python
location_sales = df.groupby("Location")["Total Spent"].sum().sort_values(ascending=False)

print("SALES BY LOCATION:")
print(location_sales)
```

```python
location_sales.plot(kind="bar", figsize=(7,5))

plt.title("Total Sales by Location")
plt.xlabel("Location")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
print(location_sales)
```

```python
df["Transaction Date"] = pd.to_datetime(df["Transaction Date"])

df["Month"] = df["Transaction Date"].dt.to_period("M")

monthly_sales = df.groupby("Month")["Total Spent"].sum()

print("MONTHLY SALES:")
print(monthly_sales)
```

```python
highest_month = monthly_sales.idxmax()
highest_month_sales = monthly_sales.max()

print("Highest Sales Month:", highest_month)
print("Total Sales:", highest_month_sales)
```

```python
print("CORRELATION BETWEEN QUANTITY AND TOTAL SPENT:")

correlation = df["Quantity"].corr(df["Total Spent"])

print(correlation)
```
