```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("data/retail_store_sales.csv")

print(df)
```

```python
print("FIRST 5 ROWS:")
df.head()
```

```python
print("DATASET SHAPE:")
print(df.shape)
```

```python
print("DATASET INFORMATION:")
df.info()
```

```python
print("MISSING VALUES:")
print(df.isnull().sum())
```

```python
print("DUPLICATE RECORDS:")
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
print("DATA TYPES:")
print(df.dtypes)
```

```python
print("UNIQUE VALUES:")

for column in df.columns:
    print("\n" + column + ":")
    print(df[column].unique()[:20])
```

```python
df["Item"] = df["Item"].fillna("Unknown")

print("Missing Item values cleaned.")
```

```python
df["Price Per Unit"] = df["Price Per Unit"].fillna(
    df["Price Per Unit"].median()
)

print("Missing Price Per Unit values cleaned.")
```

```python
df["Quantity"] = df["Quantity"].fillna(
    df["Quantity"].median()
)

print("Missing Quantity values cleaned.")
```

```python
df["Total Spent"] = df["Total Spent"].fillna(
    df["Price Per Unit"] * df["Quantity"]
)

print("Missing Total Spent values cleaned.")
```

```python
df["Discount Applied"] = df["Discount Applied"].fillna(False)

print("Missing Discount Applied values cleaned.")
```

```python
df = df.drop_duplicates()

print("Duplicate records removed.")
print("Remaining duplicates:", df.duplicated().sum())
```

```python
df["Transaction Date"] = pd.to_datetime(
    df["Transaction Date"],
    errors="coerce"
)

print("Transaction Date data type:")
print(df["Transaction Date"].dtype)
```

```python
df["Quantity"] = df["Quantity"].astype(int)

print("Quantity data type:")
print(df["Quantity"].dtype)
```

```python
df["Discount Applied"] = df["Discount Applied"].astype(bool)

print("Discount Applied data type:")
print(df["Discount Applied"].dtype)
```

```python
print("MISSING VALUES AFTER CLEANING:")
print(df.isnull().sum())
```

```python
print("DUPLICATES AFTER CLEANING:")
print(df.duplicated().sum())
```

```python
print("DATA TYPES AFTER CLEANING:")
print(df.dtypes)
```

```python
print("CLEANED DATASET:")
df.head()
```

```python
print("FINAL DATASET SHAPE:")
print(df.shape)
```

```python
df.to_csv(
    "data/retail_store_sales_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
